```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
def show_comparison(original,blur3,blur11,blur50):
  plt.figure(figsize=(12,10))
  plt.subplot(2,2,1)
  plt.imshow(original,cmap='gray')
  plt.title("Original Grayscale Image")
  plt.axis('off')

  plt.subplot(2,2,2)
  plt.imshow(blur3,cmap='gray')
  plt.title("Average Filter (3x3)")
  plt.axis('off')



  plt.subplot(2,2,3)
  plt.imshow(blur11,cmap='gray')
  plt.title("Average Filter (11x11)")
  plt.axis('off')

  plt.subplot(2,2,4)
  plt.imshow(blur50,cmap='gray')
  plt.title("Average Filter (50x50)")
  plt.axis('off')

  plt.tight_layout()
  plt.show()
```

```python
img=cv2.imread('/content/images (4).jpg',0)
if img is None:
  print("Error loading image")
else:
  kernel_3x3=np.ones((3,3),np.float32)/9

  kernel_11x11=np.ones((11,11),np.float32)/121
  kernel_50x50=np.ones((50,50),np.float32)/2500
  blur_3x3=cv2.filter2D(img,-1,kernel_3x3)

  blur_11x11=cv2.filter2D(img,-1,kernel_11x11)
  blur_50x50=cv2.filter2D(img,-1,kernel_50x50)
  show_comparison(img,blur_3x3,blur_11x11,blur_50x50)
```
